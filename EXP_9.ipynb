```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

iris = load_iris()

data = pd.DataFrame(iris.data, columns=[
'SepalLengthCm',
'SepalWidthCm',
'PetalLengthCm',
'PetalWidthCm'
])


data.insert(0, 'Id', range(1, 151))


data['Species'] = [
iris.target_names[i] for i in iris.target
]


print("First Five Rows:")
print(data.head())

print("\nDataset Information:")
data.info()


print("\nStatistical Summary:")
print(data.describe())

data.hist(figsize=(15, 10), bins=20)
plt.tight_layout()
plt.show()

sns.pairplot(
data,
hue='Species',
diag_kind='kde'
)

plt.show()
```
